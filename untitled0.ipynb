```python
from chatterbot import ChatBot
from chatterbot.trainers import ChatterBotCorpusTrainer

# 1. Criar o robô
bot = ChatBot('KarlBot') # Você pode dar o nome que quiser

# 2. Configurar o treinador
treinador = ChatterBotCorpusTrainer(bot)

# 3. Treinar com a base de dados em português
treinador.train("chatterbot.corpus.portuguese")

# 4. Loop de conversa
print("Olá! Eu sou seu chatbot. Digite 'sair' para encerrar.")

while True:
    pergunta = input("Você: ")
    if pergunta.lower() == 'sair':
        break

    resposta = bot.get_response(pergunta)
    print(f"Bot: {resposta}")
```
